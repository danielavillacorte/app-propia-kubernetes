```python
import torch
import torch.nn as nn
from torchvision.models import resnet18


def get_model(name, classes=10, pretrained=False):
    if name != "cifar_resnet110_v1":
        raise ValueError(f"Unknown model: {name}")

    weights = "DEFAULT" if pretrained else None

    model = resnet18(weights=weights)

    # Adaptar a CIFAR-10
    model.conv1 = nn.Conv2d(
        3, 64,
        kernel_size=3,
        stride=1,
        padding=1,
        bias=False
    )
    model.maxpool = nn.Identity()
    model.fc = nn.Linear(model.fc.in_features, classes)

    return model


# Crear modelo
net = get_model(
    "cifar_resnet110_v1",
    classes=10,
    pretrained=True
)

# Exportar SOLO los pesos
torch.save(net.state_dict(), "cifar_resnet110_weights.pth")

print("Pesos guardados en:")
print("cifar_resnet110_weights.pth")
```
